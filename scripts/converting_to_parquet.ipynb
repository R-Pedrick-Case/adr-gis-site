```python
hex_grid = hex_grid.set_index("id")
```

```python
ofcom_pc = gpd.read_file('/Users/pedrickr/Documents/ADR/Connectivity/gdf_merged_ofcom_pc_coverage_202507.gpkg')
ofcom_pc.head()
```

```python
ofcom_pc.columns
```

```python

# --------------------------
# 1. Select only fields needed
# --------------------------

keep_cols = [
    "postcode",
    "Gigabit availability (% premises)",
    "% of premises unable to receive 30Mbit/s",
    "% of premises below the USO",
    "geometry"
]

ofcom_map = ofcom_pc[keep_cols].copy()

# --------------------------
# 2. Create positive measures
# --------------------------

ofcom_map["Access to >=30Mbps (% premises)"] = (
    100 - ofcom_map["% of premises unable to receive 30Mbit/s"]
)

ofcom_map["Above USO (% premises)"] = (
    100 - ofcom_map["% of premises below the USO"]
)

# --------------------------
# 3. Spatial join
# --------------------------

# assuming:
# hex_grid = hex polygons
# ofcom_map = postcode centroids

joined = gpd.sjoin(
    ofcom_map,
    hex_grid,
    how="left",
    predicate="within"
)

# --------------------------
# 4. Aggregate to one hex
# --------------------------

hex_metrics = (
    joined
    .groupby("id")
    .agg(
        {
            "Gigabit availability (% premises)": "mean",
            "Access to >=30Mbps (% premises)": "mean",
            "Above USO (% premises)": "mean",
            "postcode": "count"
        }
    )
    .rename(
        columns={
            "postcode": "postcode_count"
        }
    )
    .reset_index()
)

# --------------------------
# 5. Join back to geometry
# --------------------------

hex_broadband = (
    hex_grid
    .reset_index()
    .merge(
        hex_metrics,
        on="id",
        how="left"
    )
)

# back to GeoDataFrame
hex_broadband = gpd.GeoDataFrame(
    hex_broadband,
    geometry="geometry",
    crs=hex_grid.crs
)

# --------------------------
# 6. Save for future use
# --------------------------

hex_broadband.to_parquet(
    "uk_hex_broadband_2025.parquet"
)
```
